```python
import numpy as np
```

```python
sales_data= np.array([
    [1,150000,200000,250000],
    [2,120000,160000,140000],
    [3,120000,260000,320000],
    [4,180000,240000,280000],
    [5,160000,200000,240000]
])

print("=== Analysis ===")

print("Shape : ",sales_data.shape)

print("Restaurent Yearly income : ",sales_data[:,1:])

print("Total Yearly income : ",np.sum(sales_data[:,1:],axis=0))

print("Minimum monthly income : ",np.min(sales_data[:,1:],axis=1))

print("Maximum monthly income : ",np.max(sales_data[:,1:],axis=1))

print("Average monthly income : ",np.mean(sales_data[:,1:],axis=1))

print("Cumsom monthly income : ",np.cumsum(sales_data[:,1:],axis=1))
```
