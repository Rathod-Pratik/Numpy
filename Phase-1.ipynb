<h2>Import and declare Array using numpy</h2>

```python
import numpy as np

array_1d = np.array([1,2,3])

print("1D array : ",array_1d)

array_2d = np.array([[1,2,3],[4,5,6]])
print("2D array :",array_2d)
```

<h2>Create Array from the scratch</h2>

```python
zeros = np.zeros((2,3))
print(zeros)

ones = np.ones((2,3))
print(ones)

full = np.full((2,3),5)
print(full)

random = np.random.random((2,3))
print(random)

sequence = np.arange(0,12,2)
print(sequence)
```

<h2>Array Properties</h2>

```python
arr = np.array([[1,2,3],[4,5,6]])

print("Dimension : ",arr.ndim)
print("Size : ",arr.size)
print("Shape : ",arr.shape)
print("DType : ",arr.dtype)
```

<h2>Reshape Array</h2>

```python
arr = np.arange(12)
print("Array : ",arr)

reshaped = arr.reshape((3,4))
print("Reshaped Array : ",reshaped)

flatten = reshaped.flatten()
print(flatten)

revealed = reshaped.ravel()
print(revealed)

transpose = reshaped.T
print(transpose)
```
